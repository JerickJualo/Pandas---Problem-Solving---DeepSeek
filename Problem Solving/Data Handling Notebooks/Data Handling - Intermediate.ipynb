Problem 11 – Apply a Function Row‑wise to Create a New Column
Task:
Create a new column discounted_revenue using apply() with axis=1. For each row, if the region is 'North', multiply revenue by 0.85 (15% discount); otherwise, keep revenue unchanged. Print the first 8 rows showing store_name, region, revenue, and discounted_revenue.

Pandas Concept Used:
DataFrame.apply(func, axis=1)

# Expected Output (terminal):
     store_name   region  revenue  discounted_revenue
0    Alpha Mart    North   7500.0              6375.0
1     Beta Shop    North   2250.0              1912.5
2   Gamma Store    South   1050.0              1050.0
3   Delta Depot    South   2700.0              2700.0
4  Epsilon Outlet    East   4000.0              4000.0
5   Zeta Market     East   9000.0              9000.0
6   Eta Traders     West   2500.0              2500.0
7   Theta Plaza     West    975.0               975.0

```python
def north_discount(row):
    if row['region'] == 'North':
        return row['revenue'] * 0.85
    else:
        return row['revenue']
    
store_sales['discounted_revenue'] = store_sales.apply(north_discount, axis=1)

store_sales[['store_name','region', 'revenue','discounted_revenue']].head(8)
```

Problem 12 – Use pipe() to Chain Data Cleaning Operations (revised)
Task:
Define a function clean_data(df) that:

Removes rows where revenue is less than 2000.

Sorts the DataFrame by store_name in ascending order, and within the same store, by revenue in descending order (to make the order deterministic).

Call this function on the original DataFrame using pipe(). Print the first five rows (show all columns).

Pandas Concept Used:
DataFrame.pipe(func) and custom function.

# Expected Output (terminal):
    store_id    store_name region product_category  units_sold  price_per_unit  revenue   sale_date
0      S001    Alpha Mart  North      Electronics          25           300.0   7500.0  2024-01-05
40     S001    Alpha Mart  North           Sports          50            80.0   4000.0  2024-03-02
30     S001    Alpha Mart  North             Home          30            90.0   2700.0  2024-02-16
10     S001    Alpha Mart  North         Clothing          40            50.0   2000.0  2024-01-19
41     S002     Beta Shop  North      Electronics          20           300.0   6000.0  2024-03-02

```python
def clean_data(df):
    new_df = df[df['revenue'] >= 2000]
    new_df = df.sort_values(by = ['store_name', 'revenue'], ascending=[True, False])

    return new_df

store_sales.pipe(clean_data).head(5)
```

Problem 15 – Merge Two DataFrames on a Common Column 
Task:
Create two DataFrames from the original dataset:

store_info: unique store_id and store_name (drop duplicates).

store_total_revenue: group by store_id, sum the revenue, reset index.

Merge them on store_id (inner join). Print the first 5 rows of the merged DataFrame sorted by total_revenue descending.

Pandas Concept Used:
pd.merge(), drop_duplicates(), groupby().sum()

# Expected Output (terminal):
  store_id    store_name  total_revenue
0     S008   Theta Plaza        22235.0
1     S007   Eta Traders        21105.0
2     S004   Delta Depot        20600.0
3     S010  Kappa Corner        20125.0
4     S006   Zeta Market        19190.0


```python
store_info = store_sales[['store_id', 'store_name']].drop_duplicates()

store_total_revenue = store_sales.groupby('store_id').agg(total_revenue = ('revenue','sum')).reset_index()

merge_df = store_info.merge(store_total_revenue, on='store_id', how= 'inner')

merge_df.sort_values('total_revenue', ascending=False).head(5)
```
